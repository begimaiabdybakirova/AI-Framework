```python
def mean(values):
  total = 0
  for v in values:
    total = total + v
  return total/len(values)
print(mean([10, 20, 30]))
```

```python
def my_min(values):
  m = values[0]
  for v in values:
    if v<m:
      m=v
  return m

def my_max(values):
  m = values[0]
  for v in values:
    if v>m:
      m=v
  return m
print(my_min([10, 20, 30]))
print(my_max([10, 20, 30]))
print(min([10, 20, 30]))
print(max([10, 20, 30]))

```
