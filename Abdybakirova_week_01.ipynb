```python
import matplotlib.pyplot as plt

sizes = [50, 65, 80 ,95]
prices = [180, 200, 210, 250]

plt.scatter(sizes, prices)
plt.xlabel("Size (m^2)")
plt.ylabel("Price (k)")
plt.title("Tiny housing example")
# plt.show()
```

```python
def predict_price(size):
  a = 2.0
  b = 50
  return a * size + b
print(predict_price(70))

for s, p in zip(sizes, prices):
  print(s, "real:", p, "my_guess:", predict_price(s))
```
